```python
import pandas as pd
```

```python
# rutas
path = "../data/"
```

```python
#cargar las 9 tablas
customers = pd.read_csv(path + "olist_customers_dataset.csv")
orders = pd.read_csv(path + "olist_orders_dataset.csv")
order_items = pd.read_csv(path + "olist_order_items_dataset.csv")
payments = pd.read_csv(path + "olist_order_payments_dataset.csv")
reviews = pd.read_csv(path + "olist_order_reviews_dataset.csv")
products = pd.read_csv(path + "olist_products_dataset.csv")
sellers = pd.read_csv(path + "olist_sellers_dataset.csv")
geolocation = pd.read_csv(path + "olist_geolocation_dataset.csv")
category_translation = pd.read_csv(path + "product_category_name_translation.csv")

```

```python
tablas = {
    "customers": customers, "orders": orders, "order_items": order_items,
    "payments": payments, "reviews": reviews, "products": products,
    "sellers": sellers, "geolocation": geolocation,
    "category_translation": category_translation
}

for nombre, df in tablas.items():
    print(f"{nombre}: {df.shape[0]} filas, {df.shape[1]} columnas")
```

```python
# Unir orders con customers
df= orders.merge(customers, on="customer_id", how="left")

print(f"Filas antes del merge: {orders.shape[0]}")
print(f"Filas despues del merge: {df.shape[0]}")
```

```python
# Unir df con orden_items
df = df.merge(order_items, on="order_id", how="left")

print(f"Filas antes del merge: 99441")
print(f"Filas despues del merge: {df.shape[0]}")
```

```python
# unir df con products
df = df.merge(products, on="product_id", how="left")

print(f"Filas antes: 113425")
print(f"Filas despues: {df.shape[0]}")
```

```python
# unir df con sellers
df = df.merge(sellers,  on="seller_id", how="left")

print(f"Filas antes: {df.shape[0]}")
print(f"Filas despues: {df.shape[0]}")
```

```python
# Celda 9 — payments
filas_antes = df.shape[0]          # guardamos el número ANTES de tocar df
df = df.merge(payments, on="order_id", how="left")
print(f"Filas antes: {filas_antes}")
print(f"Filas después: {df.shape[0]}")
```

```python
# Celda 10 — reviews
filas_antes = df.shape[0]
df = df.merge(reviews, on="order_id", how="left")
print(f"Filas antes: {filas_antes}")
print(f"Filas después: {df.shape[0]}")
```

```python
# Celda 11 — category_translation
filas_antes = df.shape[0]
df = df.merge(category_translation, on="product_category_name", how="left")
print(f"Filas antes: {filas_antes}")
print(f"Filas después: {df.shape[0]}")
```

```python
# revisar valores nulos por columna
nulos = df.isnull().sum()
# .isnull() marca True/False en cada celda según si está vacía
# .sum() cuenta cuántos True hay por columna (cuántos nulos tiene cada una)

nulos_porcentaje = (nulos / len(df)) * 100
# convertimos el conteo a porcentaje sobre el total de filas,
# para entender qué tan grave es cada nulo (10 nulos en 119,143 filas
# no es lo mismo que 10 nulos en 100 filas)

resumen_nulos = pd.DataFrame({
    "nulos": nulos,
    "porcentaje": nulos_porcentaje.round(2)
})
# armamos una tabla resumen con las dos columnas juntas, más fácil de leer

print(resumen_nulos[resumen_nulos["nulos"] > 0])
print(resumen_nulos[resumen_nulos["nulos"] > 0])
# filtramos para mostrar SOLO las columnas que tienen al menos un nulo —
# no tiene sentido ver las que están en 0
```

```python
# Revisar filas duplicadas
duplicados = df.duplicated().sum()
# .duplicated() marca True en cada fila que es una copia EXACTA
# de otra fila anterior (todas las columnas iguales)
# .sum() cuenta cuántas hay

print(f"Filas duplicadas: {duplicados}")
```

```python
# Revisar los tipos de datos actuales
print(df.dtypes)
```

```python
# Convertir columnas de fecha a datetime
columnas_fecha = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date",
    "shipping_limit_date",
    "review_creation_date",
    "review_answer_timestamp"
]
# Lista con los nombres de las 8 columnas que son fechas,
# identificadas en el resultado de dtypes de la celda anterior

for col in columnas_fecha:
    df[col] = pd.to_datetime(df[col], errors="coerce")
# pd.to_datetime() convierte texto a un tipo de dato "fecha" real,
# que permite hacer cálculos (restar fechas, sacar el mes, etc.)
# errors="coerce" es la parte importante: si algún valor no se puede
# convertir (texto raro, formato inesperado), en vez de romper todo
# el proceso, lo convierte en NaT (Not a Time — el "nulo" de fechas)
# Así detectamos problemas de formato sin perder el resto de la conversión

print(df[columnas_fecha].dtypes)
  
```

```python
# Confirmar que no se crearon nulos nuevos por la conversion
for col in columnas_fecha:
    print(f"{col}: {df[col].isnull().sum()} nulos")
```
